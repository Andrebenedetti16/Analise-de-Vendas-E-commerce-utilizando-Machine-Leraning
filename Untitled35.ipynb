```python
import pandas as pd
import seaborn as sns
import matplotlib as pyplot

excel_file_path = pd.read_excel('/content/Funções Básicas - Ferramenta Subtotal - Funções Básicas - Pasta de Trabalho 3.xlsx') # Caminho e nome do arquivo corrigidos
display(excel_file_path)
```

```python
df_melted = excel_file_path.melt(id_vars='Vendedor', var_name='Data', value_name='Vendas')

sales_for_vendor = df_melted.groupby('Vendedor')['Vendas'].sum()
print(sales_for_vendor)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.barplot(x=sales_for_vendor.index, y=sales_for_vendor.values)
plt.title('Vendas Vendedor')
plt.xlabel('Vendedor')
plt.ylabel('Vendas')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
